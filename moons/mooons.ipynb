```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('moons.csv')
```

```python
df.head()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data= df, x = 'X1',y= 'X2')
plt.show()
```

```python
from sklearn.cluster import DBSCAN
```

```python
model = DBSCAN(eps=0.1,min_samples=2)
```

```python
lebels = model.fit_predict(df)
lebels
```

```python
sns.scatterplot(data=df,x= 'X1',y='X2',hue=lebels,palette='rainbow')
plt.show()
```
