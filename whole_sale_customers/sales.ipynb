```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaled_x = scaler.fit_transform(df)
```

```python
from sklearn.cluster import DBSCAN
```

```python
outlire_percent = []


for eps in np.linspace(0.001,3,50):
    dbscan = DBSCAN(eps=eps,min_samples=2*scaled_x.shape[1])
    dbscan.fit(scaled_x)
    perc_outlires = 100 * np.sum(dbscan.labels_ == -1)/len(dbscan.labels_)
    outlire_percent.append(perc_outlires)
```

```python
sns.lineplot(x=np.linspace(0.001,3,50),y=outlire_percent)
plt.ylabel('percentage of points calssified as outlires')
plt.xlabel('epsilon value')
plt.show()
```
