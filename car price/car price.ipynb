```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
data = pd.read_csv('CarPrice_Assignment.csv')
```

```python
df = data.copy()
```

```python
df
```

```python
df= df.drop(['car_ID'],axis=1)
```

```python
df['CarBrand'] = df['CarName'].str.split().str[0].str.lower().str.strip()
```

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
cols = ['CarBrand']
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in cols:
    df[col] = le.fit_transform(df[col])
```

```python
df = df.drop(['CarName'],axis=1)
```

```python
df
```

```python
cols= ['fueltype','doornumber','carbody','drivewheel','enginelocation','enginetype','cylindernumber','fuelsystem']
```

```python
for col in cols:
    df[col] = le.fit_transform(df[col])
```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
numeric_cols = df.select_dtypes(include='number').columns.tolist()
numeric_cols = [col for col in numeric_cols if col != 'price']
```

```python
df[numeric_cols]=scaler.fit_transform(df[numeric_cols])
```

```python
le = LabelEncoder()
df['aspiration'] = le.fit_transform(df['aspiration'])
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = df.drop(['price'],axis=1)
y = df[['price']]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.20,shuffle=True,random_state=44)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
```

```python
reg.fit(x_train,y_train)
```

```python
y_pred = reg.predict(x_test)
```

```python
from sklearn.metrics import r2_score
```

```python
print(r2_score(y_test,y_pred))
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly = PolynomialFeatures(degree=4)
```

```python
x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.fit_transform(x_test)
```

```python
lr  = LinearRegression()
```

```python
lr.fit(x_train_poly,y_train)
```

```python
y_pred_poly = lr.predict(x_test_poly)
```

```python
print(r2_score(y_test,y_pred_poly,))
```

```python
from sklearn.linear_model import SGDRegressor
```

```python
batch_model = SGDRegressor(max_iter=150)
```

```python
batch_model.fit(x_train,y_train)
```

```python
y_pred_batch = batch_model.predict(x_test)
```

```python
print(r2_score(y_test,y_pred_batch))
```

```python
model_mini_batch = SGDRegressor(max_iter=10000,tol=1e-3)
batch_size =10

for i in range(0,x_train.shape[0],batch_size):
    X_batch = x_train[i:i + batch_size]
    y_batch = y_train[i:i + batch_size]
    
model_mini_batch.partial_fit(X_batch, y_batch)

y_pred_mini = model_mini_batch.predict(x_test)  
```

```python
print(r2_score(y_test,y_pred_mini))
```

```python
model_sgd= SGDRegressor(max_iter=100,tol=1e-3)
```

```python
model_sgd.fit(x_train,y_train)
```

```python
y_pred_sgd = model_sgd.predict(x_test)
```

```python
print(r2_score(y_test,y_pred_sgd))
```
