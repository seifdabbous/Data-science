```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
data = pd.read_csv('Bengaluru_House_Data.csv')
```

```python
df = data.copy()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['bath'].fillna(df['bath'].mode()[0],inplace=True)
```

```python
df
```

```python
df.isnull().sum()
```

```python
df['balcony'].fillna(df['balcony'].mode()[0],inplace= True)
```

```python
df.isnull().sum()
```

```python
df['size'].fillna(df['size'].mode()[0],inplace=True)
```

```python
df['size'] = df['size'].str.extract(r'(\d+)').astype(int)
```

```python
df
```

```python
df['location'].dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df = df.drop(['society','availability','location'],axis=1)
```

```python
cols = ['area_type']
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
for col in cols:
    df[col] = le.fit_transform(df[col].astype(str))
```

```python
plt.Figure(figsize=(8,6))
sns.pairplot(df)
plt.show()
```

```python
df
```

```python
df.info()
```

```python
df['total_sqft'] = pd.to_numeric(df['total_sqft'], errors='coerce')
```

```python
df.dropna(subset=['total_sqft'], inplace=True)
```

```python
x = df.drop(['price'],axis=1)
y = df[['price']]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.20,random_state=44,shuffle= True)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
```

```python
reg.fit(x_train,y_train)
```

```python
y_pred = reg.predict(x_test)
```

```python
from sklearn.metrics import r2_score
```

```python
print(r2_score(y_test,y_pred))
```
