```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('Basket_Fruits.csv')
```

```python
df = data.copy()
```

```python
transactions = df.set_index('TransactionID').stack().groupby(level=0).apply(list).tolist()
```

```python
transactions
```

```python
ohe_df = pd.get_dummies(pd.DataFrame(transactions).stack()).groupby(level=0).sum()
```

```python
ohe_df
```

```python
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

```python
frequent_itemsets = apriori(ohe_df,min_support=0.2,use_colnames=True)
```

```python
frequent_itemsets
```

```python
rules = association_rules(frequent_itemsets,metric='confidence',min_threshold=0.5)
```
