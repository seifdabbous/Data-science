```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
data = pd.read_csv('manufacturing.csv')
```

```python
df = data.copy()
```

```python
df.info()
```

```python
df.describe()
```

```python
cols =['Number of Units','Manufacturing Cost']
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
df[cols]= scaler.fit_transform(df[cols])
```

```python
df.describe()
```

```python
plt.Figure(figsize=(8,6))
sns.pairplot(df)
plt.show()
```

```python
x= df[['Number of Units']]
y=df[['Manufacturing Cost']]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_trian,x_Test,y_trian,y_test = train_test_split(x,y,test_size=0.20,random_state=44,shuffle=True )
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
```

```python
reg.fit(x_trian,y_trian)
```

```python
y_pred = reg.predict(x_Test)
```

```python
from sklearn.metrics import r2_score
```

```python
print(r2_score(y_test,y_pred))
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly = PolynomialFeatures()
```

```python
x_trian_poly = poly.fit_transform(x_trian)
x_test_poly =poly.fit_transform(x_Test)
```

```python
lr_poly= LinearRegression()
lr_poly.fit(x_trian_poly,y_trian)
```

```python
y_pred_poly = lr_poly.predict(x_test_poly)
```

```python
print(r2_score(y_test,y_pred_poly,))
```
