```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
data = pd.read_csv('social-network-ads.csv')
```

```python
df = data.copy()
```

```python
df.isnull().sum()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
df['EstimatedSalary'] = scaler.fit_transform(df[['EstimatedSalary']])
```

```python
df.head()
```

```python
plt.Figure(figsize=(8,6))
sns.histplot(df['EstimatedSalary'],kde=True)
plt.show()
```

```python
x= df.drop(['Purchased'],axis=1)
y=df[['Purchased']]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_trian,y_test=train_test_split(x,y,test_size=0.20,random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
calssifir = KNeighborsClassifier(n_neighbors=5)
calssifir.fit(x_train,y_trian)
```

```python
y_pred = calssifir.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
```

```python
cm = confusion_matrix(y_test,y_pred)
```

```python
print(cm)
```

```python
accuracy_score(y_test,y_pred)
```
